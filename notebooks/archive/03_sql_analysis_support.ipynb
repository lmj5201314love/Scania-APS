本 notebook 用于记录 SQL 支撑文件和本地辅助表产物的位置。Day 3 不做建模、不做缺失值填充、不删除字段，也不修改 `data/raw/` 原始数据。


## 1. 读取 cfg

路径、标签列、缺失值 token 和输出目录统一从 `config/config.yaml` 读取。


```python
from pathlib import Path
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

from scania_aps.config import get_config

cfg = get_config()
cfg.train_raw, cfg.test_raw, cfg.label_column, cfg.tables_dir
```

## 2. SQL 文件检查

Day 3 的 SQL 文件用于复核 Day 2 的数据规模、标签分布、缺失率和导入质量。


```python
sql_files = sorted((cfg.project_root / "sql").glob("*.sql"))
pd.DataFrame(
    [
        {"file_name": path.name, "file_path": str(path.relative_to(cfg.project_root)), "size_bytes": path.stat().st_size}
        for path in sql_files
    ]
)
```

## 3. SQL 辅助表产物检查

如果已经运行 `scripts/prepare_sql_support_tables.py`，这里可以检查供 MySQL 导入的辅助 CSV 文件。


```python
sql_support_dir = cfg.tables_dir / "sql_support"
support_files = sorted(sql_support_dir.glob("*.csv")) if sql_support_dir.exists() else []
pd.DataFrame(
    [
        {"file_name": path.name, "file_path": str(path.relative_to(cfg.project_root)), "size_bytes": path.stat().st_size}
        for path in support_files
    ]
)
```
